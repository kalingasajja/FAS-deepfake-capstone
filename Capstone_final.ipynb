Step 2: Extracting Spatial Frames

```python
import os
import json
import cv2
import numpy as np
from tqdm.notebook import tqdm

# Define paths based on our earlier setup
BASE_DIR = '/content/drive/MyDrive/Deepfake_Capstone_Spatial'
DATASET_DIR = os.path.join(BASE_DIR, 'raw_videos')
MANIFEST_PATH = os.path.join(BASE_DIR, 'manifests', 'spatial_manifest.json')
FRAMES_DIR = os.path.join(BASE_DIR, 'extracted_frames')

# Number of frames to extract per video
NUM_FRAMES = 10
# SigLIP standard input resolution
TARGET_SIZE = (224, 224)

# Load the manifest
with open(MANIFEST_PATH, 'r') as f:
    manifest = json.load(f)

def extract_frames_from_video(video_path, output_dir, video_id, num_frames=10):
    """Extracts evenly spaced frames from a video and saves them to the output directory."""
    if not os.path.exists(video_path):
        print(f"Warning: Video not found at {video_path}")
        return False

    cap = cv2.VideoCapture(video_path)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    if total_frames == 0:
        print(f"Warning: Video {video_id} has 0 frames or cannot be read.")
        cap.release()
        return False

    # Calculate exactly 10 evenly spaced frame indices
    frame_indices = np.linspace(0, total_frames - 1, num_frames, dtype=int)

    for idx, frame_idx in enumerate(frame_indices):
        # Set the frame position
        cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)
        ret, frame = cap.read()

        if ret:
            # Resize for SigLIP to save Drive space and compute during training
            frame_resized = cv2.resize(frame, TARGET_SIZE)

            # Construct the save path: e.g., video_id_frame_0.jpg
            frame_filename = f"{video_id}_frame_{idx}.jpg"
            frame_save_path = os.path.join(output_dir, frame_filename)

            cv2.imwrite(frame_save_path, frame_resized)
        else:
            print(f"Warning: Could not read frame {frame_idx} from {video_id}")

    cap.release()
    return True

# Process each split (train, val, test)
for split in ['train', 'val', 'test']:
    print(f"\nProcessing '{split}' split...")

    # Create output directories for the split and labels (0=real, 1=fake)
    split_dir = os.path.join(FRAMES_DIR, split)
    real_dir = os.path.join(split_dir, '0')
    fake_dir = os.path.join(split_dir, '1')

    os.makedirs(real_dir, exist_ok=True)
    os.makedirs(fake_dir, exist_ok=True)

    split_data = manifest.get(split, [])

    # Use tqdm for a progress bar
    for item in tqdm(split_data, desc=f"Extracting {split}"):
        video_id = item['video_id']
        rel_path = item['path']
        label = item['label']

        # Determine output folder based on label
        out_dir = fake_dir if label == 1 else real_dir

        # Absolute path to the original .mp4
        full_video_path = os.path.join(DATASET_DIR, rel_path)

        extract_frames_from_video(full_video_path, out_dir, video_id, NUM_FRAMES)

print("\nFrame extraction complete! Data is ready for the SigLIP DataLoader.")
```

Step 3:Model Architecture and Data Loaders

```python
# Install the necessary Hugging Face libraries if you haven't already
!pip install -q transformers

import torch
import torch.nn as nn
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from transformers import SiglipVisionModel

# 1. Define the Data Transformations
# SigLIP expects 224x224 images and a specific normalization standard
siglip_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])
])

# 2. Create the DataLoaders using ImageFolder
# Since we organized frames into /train/0 and /train/1, ImageFolder maps them automatically
BASE_DIR = '/content/drive/MyDrive/Deepfake_Capstone_Spatial/extracted_frames'

train_dataset = datasets.ImageFolder(root=f'{BASE_DIR}/train', transform=siglip_transform)
val_dataset = datasets.ImageFolder(root=f'{BASE_DIR}/val', transform=siglip_transform)
test_dataset = datasets.ImageFolder(root=f'{BASE_DIR}/test', transform=siglip_transform)

# Batch size of 32 or 64 is usually safe for Colab Free tier with a frozen backbone
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=2)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=2)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False, num_workers=2)

print(f"Loaded {len(train_dataset)} training images, {len(val_dataset)} validation images.")

# 3. Define the SigLIP + MLP Pipeline
class SpatialDeepfakeDetector(nn.Module):
    def __init__(self):
        super(SpatialDeepfakeDetector, self).__init__()

        # Load the pre-trained SigLIP Vision Transformer backbone
        self.siglip = SiglipVisionModel.from_pretrained("google/siglip-base-patch16-224")

        # FREEZE the SigLIP backbone to save memory and prevent destroying pre-trained weights
        for param in self.siglip.parameters():
            param.requires_grad = False

        # SigLIP 'base' outputs a 768-dimensional embedding
        hidden_size = self.siglip.config.hidden_size

        # 3-Layer MLP classification head
        self.mlp = nn.Sequential(
            nn.Linear(hidden_size, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(),
            nn.Dropout(0.3),

            nn.Linear(512, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(0.3),

            # Single output node for Binary Cross Entropy (0 = Real, 1 = Fake)
            nn.Linear(128, 1)
        )

    def forward(self, pixel_values):
        # Extract features using the frozen backbone
        outputs = self.siglip(pixel_values=pixel_values)

        # The pooler_output represents the aggregated features for the entire image
        image_embeds = outputs.pooler_output

        # Pass through the trainable MLP
        logits = self.mlp(image_embeds)
        return logits

# Initialize the model and move it to the GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = SpatialDeepfakeDetector().to(device)

print(f"Model initialized on: {device}")
```

Step 4: The Training and Validation Loop

```python
import torch.optim as optim
from torch.nn import BCEWithLogitsLoss
import copy

# Hyperparameters
EPOCHS = 5
LEARNING_RATE = 1e-3

# Loss function and Optimizer (Only optimizing the MLP head, SigLIP is frozen)
criterion = BCEWithLogitsLoss()
optimizer = optim.AdamW(model.mlp.parameters(), lr=LEARNING_RATE, weight_decay=1e-4)

# To save the best model weights
best_val_acc = 0.0
best_model_wts = copy.deepcopy(model.state_dict())
CHECKPOINT_PATH = '/content/drive/MyDrive/Deepfake_Capstone_Spatial/checkpoints/best_spatial_model.pth'

print("Starting Training Loop...")

for epoch in range(EPOCHS):
    print(f"\nEpoch {epoch+1}/{EPOCHS}")
    print("-" * 20)

    # Each epoch has a training and validation phase
    for phase in ['train', 'val']:
        if phase == 'train':
            model.train()  # Set model to training mode
            dataloader = train_loader
        else:
            model.eval()   # Set model to evaluate mode
            dataloader = val_loader

        running_loss = 0.0
        running_corrects = 0
        total_samples = 0

        # Iterate over data
        for inputs, labels in dataloader:
            inputs = inputs.to(device)
            # Reshape labels to match output logits shape: [batch_size, 1]
            labels = labels.unsqueeze(1).float().to(device)

            optimizer.zero_grad()

            # Forward pass
            # Track history only if in train phase
            with torch.set_grad_enabled(phase == 'train'):
                logits = model(inputs)
                loss = criterion(logits, labels)

                # Apply sigmoid to logits to get probabilities (0 to 1)
                preds = torch.sigmoid(logits) > 0.5

                # Backward pass and optimize only if in training phase
                if phase == 'train':
                    loss.backward()
                    optimizer.step()

            # Statistics
            running_loss += loss.item() * inputs.size(0)
            running_corrects += torch.sum(preds == labels.data)
            total_samples += inputs.size(0)

        epoch_loss = running_loss / total_samples
        epoch_acc = running_corrects.double() / total_samples

        print(f"{phase.capitalize()} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}")

        # Deep copy the model if it's the best validation accuracy we've seen
        if phase == 'val' and epoch_acc > best_val_acc:
            best_val_acc = epoch_acc
            best_model_wts = copy.deepcopy(model.state_dict())
            torch.save(model.state_dict(), CHECKPOINT_PATH)
            print(f"🌟 New best model saved with Val Acc: {best_val_acc:.4f}")

print("\nTraining complete!")
print(f"Best Validation Accuracy: {best_val_acc:.4f}")

# Load best model weights for the final testing phase later
model.load_state_dict(best_model_wts)
```

Step 5: Test split

```python
import os
import json
import torch
from PIL import Image

print("Starting video-level inference on the test set...")

# Ensure model is in evaluation mode
model.eval()

# Paths
BASE_DIR = '/content/drive/MyDrive/Deepfake_Capstone_Spatial'
FRAMES_DIR = os.path.join(BASE_DIR, 'extracted_frames', 'test')
PREDICTIONS_DIR = os.path.join(BASE_DIR, 'predictions')
PREDICTIONS_PATH = os.path.join(PREDICTIONS_DIR, 'spatial_test_preds.json')

# We'll use the manifest we loaded earlier to get the exact test video IDs and their labels
test_data = manifest.get('test', [])
predictions = {}

with torch.no_grad():
    for item in test_data:
        video_id = item['video_id']
        label = item['label']

        # Locate the frames for this specific video
        # We stored them in the '0' or '1' folder based on their true label
        folder = '1' if label == 1 else '0'
        video_frames_dir = os.path.join(FRAMES_DIR, folder)

        frame_tensors = []

        # Load all 10 frames we extracted for this video
        for idx in range(NUM_FRAMES):
            frame_filename = f"{video_id}_frame_{idx}.jpg"
            frame_path = os.path.join(video_frames_dir, frame_filename)

            if os.path.exists(frame_path):
                # Load image and apply the exact same SigLIP transformations used in training
                image = Image.open(frame_path).convert('RGB')
                tensor = siglip_transform(image)
                frame_tensors.append(tensor)
            else:
                print(f"Warning: Missing frame {frame_filename}")

        if not frame_tensors:
            print(f"Skipping {video_id} - no frames found.")
            continue

        # Stack the 10 frames into a single batch: shape [10, 3, 224, 224]
        batch_inputs = torch.stack(frame_tensors).to(device)

        # Forward pass all 10 frames through the model at once
        logits = model(batch_inputs)

        # Convert raw logits to probabilities (0.0 to 1.0)
        probabilities = torch.sigmoid(logits).cpu().squeeze()

        # Average the probabilities across the 10 frames to get the video-level score
        video_level_prob = probabilities.mean().item()

        # Save to our predictions dictionary
        predictions[video_id] = video_level_prob

# Export the predictions to JSON
with open(PREDICTIONS_PATH, 'w') as f:
    json.dump(predictions, f, indent=4)

print(f"✅ Successfully exported predictions for {len(predictions)} test videos to {PREDICTIONS_PATH}")

# Optional: Print a quick preview of the results
print("\nSample Predictions:")
for i, (vid, prob) in enumerate(predictions.items()):
    if i >= 5: break
    print(f"{vid}: {prob:.4f}")
```

```python
import json
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix

# Load your predictions (assuming they are saved in PREDICTIONS_PATH from Step 5)
BASE_DIR = '/content/drive/MyDrive/Deepfake_Capstone_Spatial'
PREDICTIONS_PATH = f'{BASE_DIR}/predictions/spatial_test_preds.json'

with open(PREDICTIONS_PATH, 'r') as f:
    preds_dict = json.load(f)

y_true = []
y_prob = []
y_pred = []

# Parse the dictionary to extract true labels and predicted probabilities
for video_id, prob in preds_dict.items():
    # If the video ID starts with 'real', label is 0, else 1
    true_label = 0 if video_id.startswith('real') else 1

    y_true.append(true_label)
    y_prob.append(prob)

    # Standard 0.5 threshold for binary classification
    y_pred.append(1 if prob >= 0.5 else 0)

# Calculate Metrics
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)
auc_roc = roc_auc_score(y_true, y_prob)
conf_matrix = confusion_matrix(y_true, y_pred)

print("--- Spatial Branch Evaluation Metrics ---")
print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")
print(f"AUC-ROC:   {auc_roc:.4f}")
print("\nConfusion Matrix (TN, FP | FN, TP):")
print(conf_matrix)

# Break down accuracy by category to highlight the face swap failure
categories = {'synthetic': {'correct': 0, 'total': 0},
              'face_swap': {'correct': 0, 'total': 0},
              'real': {'correct': 0, 'total': 0}}

for video_id, is_pred_fake in zip(preds_dict.keys(), y_pred):
    true_label = 0 if video_id.startswith('real') else 1
    is_correct = (is_pred_fake == true_label)

    if video_id.startswith('synthetic'):
        cat = 'synthetic'
    elif video_id.startswith('face_swap'):
        cat = 'face_swap'
    else:
        cat = 'real'

    categories[cat]['total'] += 1
    if is_correct:
        categories[cat]['correct'] += 1

print("\n--- Accuracy by Category ---")
for cat, stats in categories.items():
    if stats['total'] > 0:
        cat_acc = stats['correct'] / stats['total']
        print(f"{cat.capitalize()}: {cat_acc:.4f} ({stats['correct']}/{stats['total']})")
```
